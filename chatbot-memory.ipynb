```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents= "Hi my name is Alfayath. Please remember it."
)
print (response.text)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents= "what is my name "
)
print (response.text)
```

```python
conversation =[]

def chat(user_message):
    conversation.append({"role": "user", "parts": [{ "text": user_message}]})

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents= conversation
        )
        reply = response.text
    except Exception as e:
        return f"sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{ "text": reply}]})
    return reply
```

```python
chat("Hi, my name is Alfayath. Please remember it.")
```

```python
chat("What is my name?")
```

```python
for message in conversation:
    print(f"{message['role']}: {message['parts'][0]['text']}")
    print()
```
