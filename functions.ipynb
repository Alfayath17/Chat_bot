```python
def calculated_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin
```

```python
calculated_margin(100, 40)
```

```python
def calculated_profit_and_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return profit, margin

revenue = 100
expenses = 40
profit, margin = calculated_profit_and_margin(revenue, expenses)
```

```python
def calc_total_expense(*args):
    print(args)
    return sum(args)

calc_total_expense(10, 20, 30, 40, 50)
```

```python
def total_expenses(**kargs):
    total = 0 
    for key, value in kargs.items():
        total += value
    return total

total_expenses(rent=1000, utilities=200, groceries=300, transportation=150)
```
