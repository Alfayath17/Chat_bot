```python
class employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        return self.salary * 0.10

employee1 = employee("John", 50000)
print(employee1.calculate_bonus())
```

```python

from datetime import date


class Human:

    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year

    def get_age(self):
        current_year = date.today().year
        return current_year - self.birth_year

    def introduce(self):
        print(
            f"Hello, my name is {self.name}. I am {self.get_age()} years old.")


dhaval = Human("Dhaval", 1990)

print(dhaval.get_age())
dhaval.introduce()
```
