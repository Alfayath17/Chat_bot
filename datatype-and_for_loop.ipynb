```python
revenues = [1000, 2000, 3000, 4000, 5000]
expenses = [500, 1000, 1500, 2000, 2500]
type(revenues)
```

```python
financials = {
    "Q1": {" revenue": 50, "expenses": 20},
    "Q2": {" revenue": 80, "expenses": 30},
    "Q3": {" revenue": 100, "expenses": 40}
}

for quarter, fin in financials.items():
    if fin[" revenue"] < 0:
        print(f"Invalid revenue for {quarter}. Skipping.")
        continue
    margin = (fin[" revenue"] - fin["expenses"])  / fin[" revenue"] * 100
    print(f"{quarter}: {margin:.2f}%")
```
